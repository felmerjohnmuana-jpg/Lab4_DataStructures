```python
from collections import defaultdict, Counter


# Engineering Test Data Organizer

LAST_NAME = "Muana"
STUDENT_ID = "TUPM-26-4273"
SEED_NUM = 3
FAVORITE_ARTIST = "December Avenue"

test_sources = [f"{LAST_NAME}-T{i}" for i in range(1, 6)]

categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED"
]
records = []

def generate_value(seed, artist, position):
    character = artist[(position - 1) % len(artist)]
    value = (seed + position * ord(character)) % 1000 / 10
    return round(value, 2)

position = 1

for source in test_sources:
    for category in categories:
        value = generate_value(SEED_NUM, FAVORITE_ARTIST, position)
        records.append({
            "source": source,
            "category": category,
            "value": value
        })
        position += 1

records.append(records[0].copy())

print("\n======= GENERATED TEST DATA =======")
for record in records:
    print(record)

organized = defaultdict(lambda: defaultdict(list))

for record in records:
    organized[record["source"]][record["category"]].append(
        record["value"]
    )

print("\n======= ORGANIZED DATA =======")
for source in sorted(organized):
    print(f"\nSource: {source}")
    for category in sorted(organized[source]):
        print(f"  {category}: {organized[source][category]}")

record_counts = Counter(
    (r["source"], r["category"], r["value"])
    for r in records
)

repeated_records = {
    record: count
    for record, count in record_counts.items()
    if count > 1
}

print("\n======= REPEATED RECORDS =======")
for record, count in repeated_records.items():
    print(
        f"Source={record[0]}, "
        f"Category={record[1]}, "
        f"Value={record[2]}, Count={count}"
    )

distinct_records = {
    (record["source"], record["category"], record["value"])
    for record in records
}

print(f"\nTotal records: {len(records)}")
print(f"Distinct records: {len(distinct_records)}")
print(f"Duplicate occurrences beyond first: {len(records) - len(distinct_records)}")

values_by_category = defaultdict(list)
source_counts = Counter()
category_counts = Counter()

for record in records:
    values_by_category[record["category"]].append(record["value"])
    source_counts[record["source"]] += 1
    category_counts[record["category"]] += 1

print("\n======= NUMERICAL SUMMARIES =======")
print("\nRecords per source:")
for source in sorted(source_counts):
    print(f"  {source}: {source_counts[source]}")

print("\nRecords per category:")
for category in sorted(category_counts):
    print(f"  {category}: {category_counts[category]}")

ordered_records = sorted(
    records,
    key=lambda r: (r["source"], r["category"], r["value"])
)

print("\n======= ORDERED RESULTS =======")
for record in ordered_records:
    print(
        f"{record['source']} | "
        f"{record['category']} | "
        f"{record['value']}"
    )

all_values = [record["value"] for record in records]

print("\n======= FINAL SUMMARY =======")
print(f"Student last name: {LAST_NAME}")
print(f"Favorite artist: {FAVORITE_ARTIST}")
print(f"Seed Number: {SEED_NUM}")
print(f"Number of sources: {len(test_sources)}")
print(f"Number of categories: {len(categories)}")
print(f"Total Measurements: {len(records)}")
print(f"Distinct records: {len(distinct_records)}")
print(f"Repeated record types: {len(repeated_records)}")
print(f"Overall minimum: {min(all_values)}")
print(f"Overall maximum: {max(all_values)}")
print(f"Overall average: {sum(all_values) / len(all_values):.2f}")

print("\nTest data organization completed!")

```

```python
# STUDENT PERFORMANCE DATA ANALYZER


# STUDENT-SPECIFIC INPUTS

LAST_NAME = input("Muana: ").strip().upper()
SEED_NUM = int(input("3: "))
FAVORITE_ARTIST = input("December Avenue: ").strip().upper()

if not LAST_NAME or not FAVORITE_ARTIST:
    raise ValueError("Last name and favorite artist cannot be empty.")

if SEED_NUM < 0:
    raise ValueError("Seed number must be non-negative.")



# 2. SUBJECTS AND STUDENT NAMES

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

students = [
    f"{LAST_NAME}_STUDENT_{i}"
    for i in range(1, 6)
]

# PERFORMANCE CATEGORY

def get_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 85:
        return "VERY GOOD"
    elif score >= 80:
        return "GOOD"
    elif score >= 75:
        return "PASSED"
    else:
        return "NEEDS IMPROVEMENT"



# GENERATE SCORES

# Use the positions of the characters in the artist's name.
artist_value = sum(
    (position + 1) * ord(character)
    for position, character in enumerate(FAVORITE_ARTIST)
)

records = []

for student_index, student in enumerate(students):
    for subject_index, subject in enumerate(subjects):

        score = (
            SEED_NUM
            + artist_value
            + (student_index + 1) * 11
            + (subject_index + 1) * 7
        ) % 51 + 50

        records.append({
            "student": student,
            "subject": subject,
            "score": score,
            "category": get_category(score)
        })


# ADD A REPEATED RECORD

# Repeat the first student's first subject record.
duplicate = records[0].copy()
records.append(duplicate)

# ADD AN INCOMPLETE RECORD


# Give the second student a missing score in ENGLISH.
for record in records:
    if (
        record["student"] == students[1]
        and record["subject"] == "ENGLISH"
    ):
        record["score"] = None
        record["category"] = "INCOMPLETE"
        break


# IDENTIFY DUPLICATE RECORDS

seen = set()
duplicate_records = []

for record in records:
    key = (record["student"], record["subject"])

    if key in seen:
        duplicate_records.append(record)
    else:
        seen.add(key)



# IDENTIFY INCOMPLETE RECORDS

incomplete_records = [
    record for record in records
    if record["score"] is None
]


# ORGANIZE THE RECORD

records.sort(
    key=lambda record: (
        record["student"],
        record["subject"]
    )
)


# DISPLAY ALL RECORDS

print("\n" + "=" * 90)
print("             FINAL STUDENT PERFORMANCE REPORT")
print("=" * 90)

print(f"Last Name       : {LAST_NAME}")
print(f"Seed Number     : {SEED_NUM}")
print(f"Favorite Artist : {FAVORITE_ARTIST}")

print("-" * 90)

print(
    f"{'STUDENT':<25}"
    f"{'SUBJECT':<18}"
    f"{'SCORE':<12}"
    f"{'CATEGORY':<25}"
    f"{'STATUS'}"
)

print("-" * 90)

for record in records:
    key = (record["student"], record["subject"])

    if key in {
        (r["student"], r["subject"])
        for r in duplicate_records
    }:
        status = "DUPLICATE"
    elif record["score"] is None:
        status = "INCOMPLETE"
    else:
        status = "VALID"

    score_display = (
        str(record["score"])
        if record["score"] is not None
        else "MISSING"
    )

    print(
        f"{record['student']:<25}"
        f"{record['subject']:<18}"
        f"{score_display:<12}"
        f"{record['category']:<25}"
        f"{status}"
    )



# STUDENT PERFORMANCE SUMMARY

print("\n" + "=" * 90)
print("                  STUDENT PERFORMANCE SUMMARY")
print("=" * 90)

for student in students:

    student_records = [
        record for record in records
        if record["student"] == student
    ]

    # Exclude duplicates and missing scores from calculations.
    valid_scores = []

    for record in student_records:
        key = (record["student"], record["subject"])

        if (
            record["score"] is not None
            and key not in {
                (r["student"], r["subject"])
                for r in duplicate_records
            }
        ):
            valid_scores.append(record["score"])

    if valid_scores:
        average = sum(valid_scores) / len(valid_scores)
        highest = max(valid_scores)
        lowest = min(valid_scores)

        print(f"\nStudent: {student}")
        print(f"Valid Scores: {len(valid_scores)}")
        print(f"Average Score: {average:.2f}")
        print(f"Highest Score: {highest}")
        print(f"Lowest Score: {lowest}")
        print(f"Overall Category: {get_category(average)}")
    else:
        print(f"\nStudent: {student}")
        print("No valid scores available.")


# SUBJECT PERFORMANCE SUMMARY


print("\n" + "=" * 90)
print("                  SUBJECT PERFORMANCE SUMMARY")
print("=" * 90)

duplicate_keys = {
    (record["student"], record["subject"])
    for record in duplicate_records
}

for subject in subjects:

    subject_records = [
        record for record in records
        if record["subject"] == subject
    ]

    valid_scores = [
        record["score"]
        for record in subject_records
        if record["score"] is not None
        and (record["student"], record["subject"])
        not in duplicate_keys
    ]

    if valid_scores:
        average = sum(valid_scores) / len(valid_scores)

        print(f"\nSubject: {subject}")
        print(f"Number of Valid Scores: {len(valid_scores)}")
        print(f"Average Score: {average:.2f}")
        print(f"Highest Score: {max(valid_scores)}")
        print(f"Lowest Score: {min(valid_scores)}")
        print(f"Category: {get_category(average)}")


# DATA QUALITY REPORT

print("\n" + "=" * 90)
print("                     DATA QUALITY REPORT")
print("=" * 90)

print(f"Total Records: {len(records)}")
print(f"Unique Student-Subject Pairs: {len(seen)}")
print(f"Repeated Records: {len(duplicate_records)}")
print(f"Incomplete Records: {len(incomplete_records)}")

print("\nRepeated Records:")

if duplicate_records:
    for record in duplicate_records:
        print(
            f"{record['student']} - {record['subject']}"
        )
else:
    print("None")

print("\nIncomplete Records:")

if incomplete_records:
    for record in incomplete_records:
        print(
            f"{record['student']} - {record['subject']}"
        )
else:
    print("None")



# FINAL REPORT

print("\n" + "=" * 90)
print("                 END OF STUDENT PERFORMANCE REPORT")
print("=" * 90)
```

```python

# TRANSACTION HISTORY ANALYZER
# Student Inputs

LAST_NAME = "MUANA"
SEED_NUM = 3
FAVORITE_ARTIST = "DECEMBER AVENUE"

# GENERATE CUSTOMERS

customers = [f"{LAST_NAME}-C{i}" for i in range(1, 6)]

# PRODUCTS AND FIXED PRICES

products = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 800,
    "CALCULATOR": 500
}

# GENERATE QUANTITIES
# Using character positions of FAVORITE_ARTIST


artist = FAVORITE_ARTIST.replace(" ", "").upper()

positions = [ord(char) - 64 for char in artist if char.isalpha()]
quantities = [SEED_NUM * pos for pos in positions]


# TRANSACTION RECORDS
# Includes repeated purchase


transactions = [
    (customers[0], "NOTEBOOK", quantities[0]),
    (customers[1], "PEN", quantities[1]),
    (customers[2], "USB", quantities[2]),
    (customers[3], "HEADPHONES", quantities[3]),
    (customers[4], "CALCULATOR", quantities[4]),

    (customers[0], "USB", quantities[5]),
    (customers[1], "HEADPHONES", quantities[6]),
    (customers[2], "NOTEBOOK", quantities[7]),
    (customers[3], "PEN", quantities[8]),
    (customers[4], "USB", quantities[9]),

    # Repeated transaction
    (customers[0], "NOTEBOOK", quantities[0]),

    # Additional transactions
    (customers[1], "CALCULATOR", quantities[10]),
    (customers[2], "HEADPHONES", quantities[11]),
    (customers[3], "USB", quantities[12]),
    (customers[4], "NOTEBOOK", quantities[13])
]


# GENERATED TRANSACTION RECORDS

print("\n=== GENERATED TRANSACTION RECORDS ===")

for customer, product, quantity in transactions:
    value = quantity * products[product]
    print(
        f"{customer:10} | "
        f"{product:12} | "
        f"Qty: {quantity:3} | "
        f"Value: PHP {value:,.2f}"
    )


# 2. ORGANIZED TRANSACTION DATA

print("\n=== ORGANIZED TRANSACTION DATA ===")

customer_records = {}

for customer, product, quantity in transactions:
    customer_records.setdefault(customer, []).append(
        (product, quantity)
    )

for customer, data in customer_records.items():
    print(f"\n{customer}")
    for product, quantity in data:
        print(f"  {product:<12} Qty: {quantity}")

# REPEATED/DISTINCT TRANSACTIONS


print("\n=== REPEATED TRANSACTIONS ===")

count = {}
repeated = []

for trans in transactions:
    count[trans] = count.get(trans, 0) + 1

for item, freq in count.items():
    if freq > 1:
        repeated.append(item)
        print(item)

print("\nDistinct Transactions:", len(count))

# TRANSACTION VALUES

print("\n=== TRANSACTION VALUES ===")

transaction_values = []

for customer, product, quantity in transactions:
    value = quantity * products[product]
    transaction_values.append(value)

total_dataset_value = sum(transaction_values)

print("Overall Dataset Value: PHP", format(total_dataset_value, ",.2f"))

#  CUSTOMER SUMMARIES


print("\n=== CUSTOMER SUMMARIES ===")

customer_totals = {}

for customer, product, quantity in transactions:
    value = quantity * products[product]

    customer_totals[customer] = (
        customer_totals.get(customer, 0) + value
    )

for customer, total in customer_totals.items():
    print(f"{customer}: PHP {total:,.2f}")

# PRODUCT SUMMARIES

print("\n=== PRODUCT SUMMARIES ===")

product_totals = {}

for customer, product, quantity in transactions:
    value = quantity * products[product]

    product_totals[product] = (
        product_totals.get(product, 0) + value
    )

for product, total in product_totals.items():
    print(f"{product}: PHP {total:,.2f}")


# ORDERED RESULTS

print("\n=== ORDERED CUSTOMER TOTALS ===")

ordered_customers = sorted(
    customer_totals.items(),
    key=lambda x: x[1],
    reverse=True
)

for customer, total in ordered_customers:
    print(f"{customer}: PHP {total:,.2f}")

# OVERALL TRANSACTION SUMMARY

highest_customer = max(
    customer_totals,
    key=customer_totals.get
)

highest_product = max(
    product_totals,
    key=product_totals.get
)

print("\n=== OVERALL TRANSACTION SUMMARY ===")

print("Total Transactions :", len(transactions))
print("Distinct Transactions :", len(count))
print("Repeated Transactions :", len(repeated))
print("Overall Dataset Value : PHP {:,.2f}".format(total_dataset_value))
print("Top Customer :", highest_customer)
print("Top Product :", highest_product)

print("\nFINAL CONSOLIDATED TRANSACTION REPORT COMPLETE")
```
